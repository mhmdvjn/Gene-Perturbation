# Find Druggable Hypotheses

> Research question: **Which gene targets, when perturbed, suggest a druggable hypothesis?**

Deliverables:
1. Predicted transcriptional response per candidate target
2. Dependency + drug-target annotation overlay per candidate
3. Prioritized testable-target list annotated with evidence and caveat

A target hypothesis is not a validated drug.

## Setup & imports

```python
import os, sys, pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch

sys.path.insert(0, os.path.join(os.getcwd(), '..')) if os.path.isdir(os.path.join(os.getcwd(), '..', 'materials')) else None
_cwd = os.getcwd()
for _c in [os.path.join(_cwd, 'materials'), os.path.join(_cwd, '..', 'materials')]:
    if os.path.isdir(os.path.join(_c, 'gears')) and _c not in sys.path:
        sys.path.insert(0, _c)
        break

from gears import PertData, GEARS

DATASET = 'norman'
DATA_DIR = os.path.join(os.getcwd(), '..', 'data')
RESULTS_DIR = os.path.join(os.getcwd(), 'results')
MODEL_DIR = os.path.join(os.getcwd(), 'models')
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(MODEL_DIR, exist_ok=True)

device = 'mps' if torch.backends.mps.is_available() else ('cuda' if torch.cuda.is_available() else 'cpu')
print('Using device:', device)
```

## Load data & prepare simulation split

```python
pert_data = PertData(DATA_DIR)
pert_data.load(data_name=DATASET)
pert_data.prepare_split(split='simulation', seed=1)
pert_data.get_dataloader(batch_size=32, test_batch_size=128)
print('num perts:', len(pert_data.pert_names))
print('splits:', {k: len(v) for k, v in pert_data.set2conditions.items()})
print('adata shape:', pert_data.adata.shape)
```

## Extract candidate target genes

```python
def extract_single_gene_targets(conditions):
    targets = set()
    for condition in conditions:
        if not isinstance(condition, str) or condition.strip().lower() == 'ctrl':
            continue
        genes = [g.strip() for g in condition.split('+') if g.strip().lower() != 'ctrl']
        if len(genes) == 1:
            targets.add(genes[0])
    return sorted(targets)

test_conditions = pert_data.set2conditions.get('test', [])
candidate_targets = extract_single_gene_targets(test_conditions)
if not candidate_targets:
    all_split_conditions = []
    for split in ['train', 'val', 'validation', 'test']:
        all_split_conditions.extend(pert_data.set2conditions.get(split, []))
    candidate_targets = extract_single_gene_targets(all_split_conditions)
if not candidate_targets:
    candidate_targets = extract_single_gene_targets(pert_data.adata.obs['condition'].dropna().unique().tolist())

pert_upper_to_symbol = {str(g).upper(): g for g in pert_data.pert_names}
resolved, skipped = [], []
for t in candidate_targets:
    key = t.strip().upper()
    (resolved if key in pert_upper_to_symbol else skipped).append(pert_upper_to_symbol.get(key, t))
candidate_targets = resolved
print('Resolved targets:', len(candidate_targets), '| skipped:', len(skipped))
```

## Gene dependency annotation

```python
DEP_PATH = os.path.join(DATA_DIR, 'essential_all_data_pert_genes.pkl')
essential_genes = set()
if os.path.exists(DEP_PATH):
    with open(DEP_PATH, 'rb') as f:
        essential_genes = {str(g).strip().upper() for g in pickle.load(f) if str(g).strip()}
dependency_score = {t: (1.0 if t.upper() in essential_genes else 0.0) for t in candidate_targets}
n_dep = int(sum(1 for t in candidate_targets if dependency_score[t] > 0))
print('Essential targets:', n_dep, '/', len(candidate_targets))
```

## Initialize / load / train GEARS model

```python
model = GEARS(pert_data, device=device, weight_bias_track=False)
model_path = os.path.join(MODEL_DIR, 'drug')
config_path = os.path.join(model_path, 'config.pkl')
weights_path = os.path.join(model_path, 'model.pt')
loaded = False
if os.path.isdir(model_path) and os.path.exists(config_path) and os.path.exists(weights_path):
    try:
        model.load_pretrained(model_path)
        loaded = True
        print('Loaded pretrained model')
    except Exception as e:
        print('Load failed:', repr(e))
if not loaded:
    model.model_initialize(hidden_size=16)
    model.train(epochs=20, lr=1e-3, weight_decay=5e-4)
    model.save_model(model_path)
    print('Trained and saved model')
```

## Predict transcriptional response & score targets

```python
ctrl_expression = model.ctrl_expression.detach().cpu().numpy().reshape(-1)
pred_results = model.predict([[g] for g in candidate_targets])
rows, missing = [], []
for g in candidate_targets:
    pred = pred_results.get(g)
    if pred is None:
        missing.append(g); continue
    pred = np.asarray(pred, dtype=np.float64).reshape(-1)
    if pred.shape[0] != ctrl_expression.shape[0]:
        missing.append(g); continue
    rows.append({'target': g, 'score': float(np.mean(np.abs(pred - ctrl_expression)))})
score_df = pd.DataFrame(rows)
score_df['dependency_score'] = score_df['target'].map(lambda g: dependency_score.get(g, 0.0))
score_df.to_csv(os.path.join(RESULTS_DIR, 'target_scores_raw.csv'), index=False)
print('Scored:', len(score_df), '| missing:', len(missing))
```

## Drug-target annotation

```python
from drug_annotations import build_annotation_dicts
drug_targets, drug_status = build_annotation_dicts()
DRUG_PKL = os.path.join(DATA_DIR, 'drug_targets.pkl')
with open(DRUG_PKL, 'wb') as f:
    pickle.dump({'drug_targets': drug_targets, 'drug_status': drug_status}, f)

score_df['drug_coverage'] = score_df['target'].map(lambda t: drug_status.get(str(t).strip().upper(), 'no_coverage'))
score_df['associated_drugs'] = score_df['target'].map(lambda t: drug_targets.get(str(t).strip().upper(), []))
print(score_df['drug_coverage'].value_counts())
```

## Rank candidates

```python
ranked = score_df.sort_values('score', ascending=False).reset_index(drop=True)
ranked['rank'] = ranked.index + 1
ranked.to_csv(os.path.join(RESULTS_DIR, 'ranked_targets.csv'), index=False)
print(ranked.head(20).to_string(index=False))
```

## Summary report

```python
report_lines = [
    '# Find Druggable Hypotheses — Results Summary',
    '',
    f'Dataset: {DATASET}',
    f'Candidate targets evaluated: {len(ranked)}',
    f'Targets with dependency annotation: {n_dep}',
    '',
    '## Top 10 ranked candidate targets',
    '',
    '| Rank | Target | Score | Dependency | Drug coverage | Associated drugs |',
    '|------|--------|-------|------------|---------------|------------------|',
]
for _, r in ranked.head(10).iterrows():
    drugs = ', '.join(r['associated_drugs']) if r['associated_drugs'] else '—'
    report_lines.append(f"| {int(r['rank'])} | {r['target']} | {r['score']:.4f} | {r['dependency_score']:.4f} | {r['drug_coverage']} | {drugs} |")
report_lines += ['', '## Caveat', '', 'A target hypothesis is not a validated drug. Predicted response magnitude and dependency evidence prioritize testable targets; experimental validation is required before any drug claim.', '', '## Files produced', '- target_scores_raw.csv', '- ranked_targets.csv', '- top_targets_barplot.png', '- score_distribution.png']
report = '\n'.join(report_lines)
print(report)
with open(os.path.join(RESULTS_DIR, 'interpretation_report.md'), 'w') as f:
    f.write(report)
```
